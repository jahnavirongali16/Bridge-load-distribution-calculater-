```python
import numpy as np
import matplotlib.pyplot as plt
```

```python

L = float(input("Enter bridge length L (m): "))
P = float(input("Enter load P (kN): "))
x = float(input("Enter distance of load from left support x (m): "))

if L <= 0:
    print("Error: Bridge length must be greater than 0.")

elif P < 0:
    print("Error: Load cannot be negative.")

elif x < 0 or x > L:
    print("Error: Load position must be between 0 and the bridge length.")

else:
    print("\nInputs accepted.")
```

```python
if L > 0 and P >= 0 and 0 <= x <= L:

    A = np.array([
        [1, 1],
        [0, L]
    ])

    B = np.array([
        P,
        P * x
    ])

    reactions = np.linalg.solve(A, B)

    RA = reactions[0]
    RB = reactions[1]

    print("Matrix A:")
    print(A)

    print("\nMatrix B:")
    print(B)

    print("\nSupport reactions:")
    print(f"RA = {RA:.2f} kN")
    print(f"RB = {RB:.2f} kN")
```
